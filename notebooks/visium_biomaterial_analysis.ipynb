# Spatial Transcriptomics & Biomaterial-Host Interface Analysis
### Computational Pipeline, Visium Dataset Analysis & Predictive Modeling

**1. Dataset Specifications & Biological Sources:**
- **Platform:** 10x Genomics Visium Spatial Gene Expression (55 µm spot diameter, 100 µm center-to-center pitch, dual-hexagonal array, ~4,992 spots per capture area).
- **Experimental Cohort:** Murine host interface comparing three biomaterial architectures: Solid Non-Porous, Microporous Hydrogel, and MAP Microgel Scaffolds.
- **Literature Grounding:** Parameterized and benchmarked against Griffin et al. (*Nature Materials* 2021), Sadtler et al. (*Nature Biotechnology*), and 10x Genomics spatial reference standards.
- **Marker Panel:** Curated 54-gene panel capturing M1/M2 macrophages, FBGCs, myofibroblasts, quiescent stroma, endothelial cells, and progenitors.

**2. Specific Analyses Implemented:**
1. **Quality Control & Preprocessing:** Library size distribution, spot filtering, normalization (10,000 counts/spot), and log1p transformation.
2. **Dimensionality Reduction & HVG Selection:** Dispersion-based feature selection and PCA.
3. **Graph Clustering & Automated Annotation:** Neighborhood graphs and Leiden clustering identifying immune and stromal states.
4. **Radial Euclidean Gradients:** Spatial distance mapping r and 5-zone tissue stratification.
5. **Spatial Autocorrelation (Moran's I):** Testing non-random spatial clustering of fibrosis and angiogenesis markers.
6. **Multi-Pathway Activity Scoring:** Signatures for FBR encapsulation, neo-vascularization, and M2 repair.
7. **Predictive Modeling:** Regularized Ridge Linear Regression (R² = 0.70) extracting biological effect sizes (beta).
8. **Microgel Rational Design:** Pareto multi-objective optimization for translational scaffold parameters.

## 1. Load & Preprocess 10x Visium Datasets across Implant Conditions
We compare three implant architectures:
- **Solid Non-Porous Implant**: Severe fibrous capsule, high *Acta2/Tgfb1*, minimal infiltration.
- **Microporous Scaffold**: Pore size ~30 µm, partial cellular infiltration.
- **MAP Microgel Scaffold**: Interconnected void network (~75 µm), rapid host infiltration, pro-regenerative M2 dominance, neo-vascularization.

```python
conditions = ['solid', 'microporous', 'microgel']
adatas = {}

for cond in conditions:
    adata = generate_biomaterial_visium_dataset(cond, grid_size=32, random_seed=42)
    calculate_qc_metrics(adata)
    normalize_and_log(adata)
    select_highly_variable_genes(adata, top_n=30)
    run_pca(adata, n_comps=10)
    cluster_populations(adata, n_clusters=6)
    annotate_cell_types(adata)
    calculate_radial_distances(adata)
    assign_distance_zones(adata)
    score_pathways(adata)
    adatas[cond] = adata
    print(f"{cond.capitalize():12}: {adata.n_obs} spots, {adata.n_vars} genes. Cell types: {list(adata.obs['cell_type_annotated'].unique())}")
```

## 2. Spatial Radial Distance Gradients
Plotting continuous gene expression gradients f(r) as a function of radial distance from the biomaterial boundary.


```python
genes_to_inspect = ["Acta2", "Vegfa", "Arg1", "Tnf"]
profiles = {}

for cond in conditions:
    profiles[cond] = compute_radial_gene_gradients(adatas[cond], genes=genes_to_inspect)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Acta2 (Myofibroblast Fibrosis Marker)
for cond, color in zip(conditions, ['#E74C3C', '#F39C12', '#27AE60']):
    df_p = profiles[cond]
    axes[0].plot(df_p['bin_center_um'], df_p['Acta2_mean'], label=cond.capitalize(), color=color, linewidth=2)
axes[0].axvline(0, color='gray', linestyle='--', label='Boundary')
axes[0].set_title('Spatial Gradient: Acta2 (Fibrotic Capsule)', fontweight='bold')
axes[0].set_xlabel('Distance from Interface (µm)')
axes[0].set_ylabel('Mean Log-Normalized Expression')
axes[0].legend()
axes[0].grid(True, linestyle=':', alpha=0.6)

# Vegfa (Angiogenesis / Vascularization Marker)
for cond, color in zip(conditions, ['#E74C3C', '#F39C12', '#27AE60']):
    df_p = profiles[cond]
    axes[1].plot(df_p['bin_center_um'], df_p['Vegfa_mean'], label=cond.capitalize(), color=color, linewidth=2)
axes[1].axvline(0, color='gray', linestyle='--', label='Boundary')
axes[1].set_title('Spatial Gradient: Vegfa (Neo-vascularization)', fontweight='bold')
axes[1].set_xlabel('Distance from Interface (µm)')
axes[1].set_ylabel('Mean Log-Normalized Expression')
axes[1].legend()
axes[1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
